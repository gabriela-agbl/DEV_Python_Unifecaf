# 5. Shape e TypedDict

Um `dict` genérico pode comunicar pouco sobre os campos esperados.

Compare:

```python
usuario: dict[str, object]
```

com um shape explicitamente documentado.

## Heurística SHAPE

- **S — Semantics:** o que este dado significa?
- **H — How structured:** como ele está organizado?
- **A — Allowed values:** que tipos/valores esperamos?
- **P — Presence:** quais campos precisam existir?
- **E — Expectations:** como o dado será usado?

> `SHAPE` aqui é uma heurística pedagógica da aula, não uma palavra-chave oficial do Python.

# 7. Runtime Trap — o mundo externo não lê seus type hints

## NÃO É ENTREGÁVEL SEPARADO

Agora vamos observar o limite mais importante da aula.

Imagine que este dicionário veio de uma API, arquivo ou JSON externo:

````python
from typing import TypedDict


class PedidoRuntime(TypedDict):
    total: float


pedido_externo = {
    "total": "100.00"
}

print("Valor recebido:", pedido_externo["total"])
print("Tipo real:", type(pedido_externo["total"]))
````
